```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("data.csv", na_values="NA")

# Plot heights
plt.figure()
plt.hist(data["height"].dropna(), bins=10, edgecolor="black")
plt.title("Height Distribution")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# Plot icecream
plt.figure()
data[data["flavour"].notnull()].groupby("flavour").size().plot(kind="pie")
counts = data["flavour"].dropna().value_counts()
counts.plot(kind="pie", autopct="%1.1f%%", labels=None) 

plt.title("Favourite Ice Cream Flavours")
plt.ylabel("")
plt.legend(labels=counts.index,title="Flavour",loc="best")
plt.tight_layout()
plt.show()
```
